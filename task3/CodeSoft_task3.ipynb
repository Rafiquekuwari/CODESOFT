```python
import torch
from transformers import BlipProcessor, BlipForConditionalGeneration
from google.colab import files
import matplotlib.pyplot as plt
from PIL import Image
```

```python
# Load pre-trained BLIP model and processor
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base")
```

```python
def generate_caption_from_uploaded_image():
    # Upload image from local machine
    uploaded = files.upload()

    for img_path in uploaded.keys():
        # Open and display the uploaded image
        img = Image.open(img_path)
        plt.imshow(img)
        plt.axis('off')  # Remove axes for a cleaner look
        plt.show()

        # Preprocess image and use BLIP model to generate caption
        inputs = processor(images=img, return_tensors="pt")
        out = model.generate(**inputs)

        # Decode the generated caption
        caption = processor.decode(out[0], skip_special_tokens=True)
        print("Generated Caption: ", caption)
```

```python
generate_caption_from_uploaded_image()
```

 2: Import Libraries

```python
import torch
from transformers import BlipProcessor, BlipForConditionalGeneration
from PIL import Image
import gradio as gr
```

3: Load BLIP Model and Processor

```python
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base")
device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)
```

4: Captioning Function

```python
def generate_caption(image):
    image = image.convert("RGB")
    inputs = processor(images=image, return_tensors="pt").to(device)

    out = model.generate(**inputs)
    caption = processor.decode(out[0], skip_special_tokens=True)
    return caption
```

5: Gradio Interface

```python
interface = gr.Interface(
    fn=generate_caption,
    inputs=gr.Image(type="pil"),
    outputs="text",
    title="🖼️ AI Image Captioning with BLIP",
    description="Upload an image and let the AI describe it!",
    allow_flagging="never"
)

interface.launch()
```
